```python
import seaborn as sns
sns.get_dataset_names
df=sns.load_dataset("tips")
df.head()
```

```python
# tips average
sns.boxplot(data=df,y="tip")
```

```python
#  bill amount and tip relation
sns.scatterplot(data=df,x="total_bill",y="tip",hue="sex")
```

```python
sns.scatterplot(data=df,x="total_bill",y="tip",hue="smoker")
```

```python
sns.scatterplot(data=df,x="total_bill",y="tip",hue="time")
```

```python
# bill amount distribution

sns.histplot(data=df,x="total_bill",color="green")
```

```python
cl=["total_bill","tip","size"]
sns.heatmap(df[cl].corr(),annot=True)
```

```python
sns.regplot(data=df,x="total_bill",y="tip")
```

```python
sns.lineplot(data=df,x="total_bill",y="tip",hue="sex")
```

```python
#day wise  average revenue
import numpy as np
sns.barplot(data=df,x="day",y="total_bill",color="green",estimator=np.sum)
```

```python
#day wise total bill,outliers
sns.boxplot(data=df,x="day",y="total_bill")
```

```python
sns.boxplot(data=df,x="day",y="total_bill",hue="sex")
```

```python
# tip effection factors
cl=["total_bill","tip","size"]
sns.heatmap(df[cl].corr(),annot=True)
```
