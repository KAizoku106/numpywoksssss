```python
x=df[['brand','year', 'km_driven', 'fuel', 'seller_type',
       'transmission', 'owner', 'mileage', 'engine', 'max_power', 'torque',
       'seats']]
y=df[[ 'selling_price',]]
```

```python
from sklearn.impute import SimpleImputer
si=SimpleImputer(missing_values=np.nan,strategy="mean")
x[[ 'mileage', 'engine', 'max_power', 'torque',
       'seats']]= si.fit_transform(x[[ 'mileage', 'engine', 'max_power', 'torque',
       'seats']])

```

```python
top_brands= df["brand"].value_counts().head(15).index
```

```python
x["brand"]=df["brand"].apply(lambda x:x if x in top_brands else "others")
```

```python
#encoding
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer([("encoder",OneHotEncoder(),['brand', 'fuel', 'seller_type',
       'transmission', 'owner'])],remainder='passthrough')
x=pd.DataFrame(ct.fit_transform(x))
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest= train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
li=LinearRegression()
li.fit(xtrain,ytrain)
# print(f"slope {li.coef_}")
# print(f"c-intecept  {li.intercept_}")
```

```python
ypred=li.predict(xtest)
```

```python
from sklearn.metrics import r2_score
print(f"r2_score  {r2_score(ytest,ypred)}")
```

```python
print("Train R2:", li.score(xtrain, ytrain))
print("Test R2:", li.score(xtest, ytest))
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor( random_state=42)
rf.fit(xtrain, ytrain)
ypred= rf.predict(xtest)
print(f"r2_score {r2_score(ytest,ypred)}")
print("Train R2:", rf.score(xtrain, ytrain))
print("Test R2:", rf.score(xtest, ytest))
```
