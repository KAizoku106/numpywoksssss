```python
x= df.iloc[:,1:-1]
y=df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest= train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(xtrain,ytrain)
y_pred= lr.predict(xtest)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,root_mean_squared_error
print(f"r2_score = {r2_score(ytest,y_pred)*100}")
print(f"mean_absolute_error = {mean_absolute_error(ytest,y_pred)}")
print(f"mean_squared_error = {mean_squared_error(ytest,y_pred)}")
print(f"root_mean_squared_error = {root_mean_squared_error(ytest,y_pred)}")
```
