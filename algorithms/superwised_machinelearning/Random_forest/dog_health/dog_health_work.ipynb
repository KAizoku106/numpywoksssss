```python
#combile symptoms
symptoms_columns = ["Symptom 1","Symptom 2","Symptom 3","Symptom 4","Symptom 5"]
df["All_symptoms"]= df[symptoms_columns].values.tolist()
df.drop(columns=symptoms_columns,inplace=True)
```

# multilabelbinarizer

```python
from sklearn.preprocessing import MultiLabelBinarizer
mlb= MultiLabelBinarizer()
symptom_encoded = mlb.fit_transform(df["All_symptoms"])
df.drop(columns="All_symptoms",inplace=True)
```

```python
symptom_encoded.dtype
```

```python
#convert symptom_encode into data frame
symptom_encoded_df = pd.DataFrame(symptom_encoded,columns=mlb.classes_,index=df.index)
```

```python
df= pd.concat([df,symptom_encoded_df],axis=1)
```

```python
df.head()
```

```python
df.drop(columns=["Precautions","Treatment","Emergency"],inplace=True)
```

```python
df["final_Disease"]=df["Disease"]
df.drop(columns="Disease",inplace=True)
df.head()
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
x.head()
```

```python
y.head()
```

```python
x["Dog Breed"].value_counts()
```

```python
#one hot encoding
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer([("encoder",OneHotEncoder(),["Dog Breed"])],remainder="passthrough")
x=pd.DataFrame(ct.fit_transform(x))
```

```python
x.head()
```

```python
#label encoding
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y=le.fit_transform(y)
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test =scaler.transform(x_test)
```

```python
from imblearn.over_sampling import SMOTE
sm = SMOTE(random_state=42)
x_train_smote,y_train_smote = sm.fit_resample(x_train,y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(random_state=4)
rf.fit(x_train_smote,y_train_smote)
```

```python
y_pred = rf.predict(x_test)
```

```python
from     sklearn.metrics import accuracy_score
print(f"accuracy_score ={accuracy_score(y_test,y_pred)*100}")
```

```python
from sklearn.ensemble import AdaBoostClassifier,GradientBoostingClassifier
ada= AdaBoostClassifier()
gra = GradientBoostingClassifier()
ada.fit(x_train_smote,y_train_smote)
y_pred_ada= ada.predict(x_test)
print(f"accuracy_score ={accuracy_score(y_test,y_pred_ada)*100}")
```

```python
gra.fit(x_train_smote,y_train_smote)
y_pred_gra= gra.predict(x_test)
print(f"accuracy_score ={accuracy_score(y_test,y_pred_ada)*100}")
```
