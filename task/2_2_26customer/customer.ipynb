```python
copy= df.copy()
```

```python
copy.head(2)
```

```python
copy["JoinDate"]=pd.to_datetime(copy["JoinDate"])
```

```python
copy["Join_date"]=copy["JoinDate"].dt.day
copy["Join_month"]=copy["JoinDate"].dt.month
copy["Join_year"]=copy["JoinDate"].dt.year
```

```python
copy.head(2)
```

```python
copy.drop(columns="JoinDate",inplace=True)
```

```python
copy["Age"].fillna(copy["Age"].median(),inplace=True)
```

```python
copy["Salary"].fillna(copy["Salary"].mean(),inplace=True)
```

```python
copy["Gender"].fillna("Unknown",inplace=True)
```

```python
copy["SatisfactionScore"].fillna(copy["SatisfactionScore"].mode()[0],inplace=True)
```

```python
copy.isnull().sum() 
```

```python
copy["Age"]=copy["Age"].astype("int")
```

```python
copy["Salary"]=copy["Salary"].astype("int")
```

```python
sns.boxplot(copy["Salary"])
```

```python
copy.head()
```

```python
conditions = [
    (copy["Age"] >= 18) & (copy["Age"] <= 25),
    (copy["Age"] >= 26) & (copy["Age"] <= 35),
    (copy["Age"] >= 36) & (copy["Age"] <= 50),
    (copy["Age"] > 50)  
]

choices = [
    "18-25",
    "26-35",
    "36-50",
    "50+"
]


copy["age_group"] = np.select(conditions, choices, default="Under 18")
```
