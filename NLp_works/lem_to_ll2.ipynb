```python

s1 ="Machine lerning helps computers learns patterNs from data"

s2= "Data science uses machines learning to analyze the data"

```

```python
import numpy as np
import pandas as pd
import nltk
```

# lowercasing

```python
s1= s1.lower()
s2 = s2.lower()
```

# vectorization

```python
token1 =nltk.word_tokenize(s1)
token2 =nltk.word_tokenize(s2)
```

# stop words removal


```python
from nltk.corpus import stopwords
#nltk.download("stopwords")
sw = stopwords.words("english")

s1= [w for w in token1 if w not in sw]
s2= [w for w in token2 if w not in sw]
```

```python
print(s1)
print(s2)
```

# lemmatization

```python
from nltk.stem  import WordNetLemmatizer
lm = WordNetLemmatizer()
root_words = [s1,s2]
sentences_new= []

for sentences in root_words:
    lemmas=[]
    for word in sentences:
        lemmas.append(lm.lemmatize(word))
    sentences_new.append(" ".join(lemmas))  

print(sentences_new)      

```

# importing tfidfvectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vect = TfidfVectorizer()
num_value = vect.fit_transform(sentences_new)
(num_value)
```

# check similarity by cosinesimilarity

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(num_value[0],num_value[1])
print(similarity)
```
