```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
sentence = "the# jupyter$ is@ the second@ largest@@ planet"
removal_special = re.sub("[^a-zA-Z0-9 ]","",sentence)
print(removal_special)
tokens_for_sent =nltk.word_tokenize(removal_special)
print(tokens_for_sent)
filtered_tokens1 = [w for w in tokens_for_sent if w not in sw]
filtered_sentence1 = " ".join(filtered_tokens1)
print(filtered_sentence1)
```

```python
# nltk > util > ngrams 
from nltk.util import ngrams

NGRAM = ngrams(filtered_tokens1, n=1)

lst_ngram = list(NGRAM)

print(lst_ngram)
```

```python
# stemming 
# nltk > stem > Porterstemmer ,SnowballStemmer

from nltk.stem import PorterStemmer

ps = PorterStemmer()

for t in lst_ngram:
    
    print(f"{t[0]} : {ps.stem(t[0])}")
```

```python
from nltk.stem import SnowballStemmer

sb = SnowballStemmer("english")

for t in lst_ngram:
    
    print(f"{t[0]} : {sb.stem(t[0])}")
```
