```python
import numpy as np
"""
s1=[0.03,0,-0.25,0.03,0,0.03,0.03,0,0]
s2=[0,0,-0.02,0,0,0,0,0.6,0.6]"""
```

```python
s1 = np.array([[0.03,0,-0.25,0.03,0,0.03,0.03,0,0]])
s2= np.array([0,0,-0.02,0,0,0,0,0.6,0.6])
```

```python
#dot product 
dot_prod =  np.dot(s1,s2)
print(dot_prod)
```

```python
#magnitude

m_s1=np.linalg.norm(s1)
m_s2=np.linalg.norm(s2)

print(m_s1)
print(m_s2)
```

```python
cosine_similarity = dot_prod/(m_s1*m_s2)
cosine_similarity
```
