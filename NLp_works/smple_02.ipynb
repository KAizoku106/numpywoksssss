```python
s1 = "Football worldcup is conducted by fifa every four years "
s2 = "Germany has won worldcup four times "
```

```python
import nltk
from nltk.corpus import stopwords
sw = stopwords.words("english")
sw
```

```python
s1=s1.lower()
s2=s2.lower()
s1
```

```python
# tokenization
tokens_1= nltk.word_tokenize(s1)
tokens_2=nltk.word_tokenize(s2)
print(tokens_1)
```

```python
#remoal stop words
s1_new = [word for word in tokens_1 if word not in sw]
s2_new= [word for word in tokens_2 if word not in sw]

s1_new
```

```python
#ngrma
 
from nltk.util import ngrams

NGRAM_s1 = ngrams(s1_new, n=1)
NGRAM_s2 = ngrams(s2_new, n=1)


lst_ngram1 = list(NGRAM_s1)
lst_ngram2 = list(NGRAM_s2)

print(lst_ngram1)
```

```python
# stemming
from nltk.stem import PorterStemmer
ps =PorterStemmer()
for t in lst_ngram1:
    print(f"{t[0]} : {ps.stem(t[0])}")
for t in  lst_ngram2:
    print(f"{t[0]} : {ps.stem(t[0])}")
```

```python
from nltk.stem import WordNetLemmatizer



lm = WordNetLemmatizer()

root_words= [s1_new,s2_new]
sentence_new = []

for i in root_words:
    lemmas=[]
    for j in i:
        lemmas.append(lm.lemmatize(j))
    sentence_new.append(" ".join(lemmas))

sentence_new      
# lemmas1=[]
# lemmas2 = []

# for i in s1_new:
#     lemmas1.append(lm.lemmatize(i))

# for i in s2_new:
#     lemmas2.append(lm.lemmatize(i)) 

# print(lemmas1)
# print(lemmas2)       
```

# TF idf vectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vect = TfidfVectorizer()
num_values = vect.fit_transform(sentence_new) # transform convertinto numercal values

num_values
```
